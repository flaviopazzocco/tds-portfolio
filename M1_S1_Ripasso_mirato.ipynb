## 1 · La mutabilità, e la trappola

Una lista è *mutable*: si modifica sul posto, senza crearne una nuova.
Nessuna sorpresa qui — esegui e prosegui.

```python
punteggi = [72, 81, 90]
punteggi[0] = 75
punteggi.append(95)
print(punteggi)
```

Ora la parte che conta.

> ### PREVEDI
> Cosa stampa la cella qui sotto? `[1, 2, 3]` oppure `[99, 2, 3]`?
> Decidi **prima** di eseguire.

```python
a = [1, 2, 3]
b = a
b[0] = 99

print("a vale:", a)
```

Se ti aspettavi `[1, 2, 3]`, benvenuto nel club — ed è proprio l'abitudine che
ti porti da **Matlab**, dove `b = a; b(1) = 99;` lascia `a` intatto perché Matlab copia.

In Python `b = a` **non copia niente**: crea un secondo nome per lo stesso oggetto.
La cella qui sotto lo dimostra: `id()` è l'indirizzo dell'oggetto in memoria.

```python
a = [1, 2, 3]
b = a
c = a.copy()

print("a e b sono lo stesso oggetto?", a is b, "  id uguali:", id(a) == id(b))
print("a e c sono lo stesso oggetto?", a is c, "  id uguali:", id(a) == id(c))
```

Quindi per copiare davvero: `a.copy()`, oppure `list(a)`, oppure `a[:]`.

```python
a = [1, 2, 3]
b = a.copy()
b[0] = 99

print("a:", a)
print("b:", b)
```

> ### TOCCA A TE
> La funzione qui sotto ha un bug che ora dovresti saper vedere: promette di
> restituire una lista ordinata, ma **modifica anche quella che le hai passato**.
> Eseguila così com'è per vedere il danno, poi correggila (una riga) e riesegui.

```python
def ordina(valori):
    valori.sort()          # <-- ordina SUL POSTO la lista originale
    return valori

originale = [3, 1, 2]
ordinata = ordina(originale)

print("ordinata: ", ordinata)
print("originale:", originale, "  <-- doveva restare [3, 1, 2]")
```

Nota a margine che ti servirà: `sorted(valori)` **restituisce una lista nuova**,
mentre `valori.sort()` modifica sul posto e restituisce `None`.
È la stessa distinzione, in versione libreria standard.

## 2 · Scegliere il contenitore giusto

Chi arriva dagli script mette tutto in liste, perché la lista assomiglia a un array.
Python ne ha quattro, e la scelta cambia leggibilità **e** velocità.

| tipo | quando |
|---|---|
| `list` | l'ordine conta e il contenuto cambia |
| `tuple` | l'ordine conta ma il record deve restare fisso |
| `dict` | conta il nome, non la posizione |
| `set` | contano unicità e appartenenza |

La differenza pratica più grande è tra lista e dizionario:

```python
# per posizione: devi ricordarti che il terzo elemento sono le ore
studente_lista = ["Ada", "Data Science", 5]
print(studente_lista[2])

# per nome: il codice dice cosa sta leggendo
studente_dict = {"nome": "Ada", "track": "Data Science", "ore": 5}
print(studente_dict["ore"])
```

> ### PREVEDI
> `strumenti` contiene `"python"` due volte. Quanti elementi avrà il set?

```python
strumenti = {"python", "git", "python", "colab"}

print(strumenti)
print("quanti:", len(strumenti))
print("c'e' python?", "python" in strumenti)
print("c'e' sql?   ", "sql" in strumenti)
```

Ora la parte che vale la pena vedere con i tuoi occhi invece che leggerla.

Cercare un elemento in una **lista** costa quanto scorrerla tutta.
Cercarlo in un **set** o in un **dict** costa quanto una singola operazione,
indipendentemente da quanto è grande. È lo stesso motivo per cui in un database
si mettono gli indici.

Su un milione di elementi, la differenza si misura:

```python
import time

n = 1_000_000
lista = list(range(n))
insieme = set(lista)
target = n - 1                      # il caso peggiore: l'ultimo elemento

t0 = time.perf_counter()
trovato = target in lista
t_lista = time.perf_counter() - t0

t0 = time.perf_counter()
trovato = target in insieme
t_set = time.perf_counter() - t0

print(f"ricerca nella lista: {t_lista*1000:8.3f} ms")
print(f"ricerca nel set:     {t_set*1000:8.3f} ms")
print(f"il set e' circa {t_lista/t_set:,.0f} volte piu' veloce")
```

## 3 · `print` non e' `return`

Chi scrive script per sé stampa e basta, perché guarda l'output a schermo
e non deve passare niente a nessuno. Poi un giorno quel valore serve.

> ### PREVEDI

```python
def mostra_totale(valori):
    print(sum(valori))          # mostra soltanto

risultato = mostra_totale([1, 2, 3])

print("risultato vale:", risultato)
print("di che tipo e':", type(risultato))
```

`None`. Non è un errore, è un comportamento diverso — e diventa un bug

**La regola operativa:** una funzione stampa solo se il suo lavoro è mostrare qualcosa.
In tutti gli altri casi restituisce, e la stampa la fa chi la chiama.

## 4 · Default e ritorno multiplo

Due cose che il libro usa senza spiegarle e che valgono mezza pagina di teoria.

```python
def stato_settimanale(sessioni, obiettivo=120):
    totale = sum(sessioni)
    stato = "costante" if totale >= obiettivo else "in crescita"
    return totale, stato

print(stato_settimanale([45, 30, 60]))        # obiettivo di default
print(stato_settimanale([45, 30, 60], 200))   # obiettivo passato a mano
```

- **Parametro con default** (`obiettivo=120`): chi chiama può ometterlo. Ti evita
  di scrivere due funzioni quasi identiche.
- **Ritorno multiplo** (`return totale, stato`): Python restituisce una tupla,
  e la si spacchetta direttamente.

```python
totale, stato = stato_settimanale([45, 30, 60])

print("totale:", totale)
print("stato: ", stato)
```

## 6 · File e percorsi

Quattro abitudini, tutte in una cella sola.

```python
from pathlib import Path

cartella = Path("dati")
cartella.mkdir(exist_ok=True)          # non protesta se esiste gia'
percorso = cartella / "note.txt"       # la barra non e' una divisione

with open(percorso, "w", encoding="utf-8") as f:      # 'with' chiude da solo
    f.write("python\ngit\npython\n")

with open(percorso, "r", encoding="utf-8") as f:
    testo = f.read()

righe = testo.strip().split("\n")
print("percorso:", percorso)
print("righe:   ", righe)
print("totali:", len(righe), " uniche:", len(set(righe)))
```

Su `encoding="utf-8"`: il libro non insiste, io sì. Su Windows il default non è UTF-8,
e il giorno che il file contiene una «à» o un «€» ti ritrovi caratteri corrotti.
Due parole che ti risparmiano un pomeriggio.

E davanti a un `FileNotFoundError`, la prima mossa è sempre questa — soprattutto

```python
from pathlib import Path

print("sto girando da:", Path(".").resolve())
print("qui dentro c'e':", [p.name for p in Path(".").iterdir()][:10])
```
